```python
# Missing Indicator + Random Sample Imputation [ annual_income ]

# BEFORE
print("Missing values BEFORE:")
print(data['annual_income'].isnull().sum())

# - Missing Indicator
data['annual_income_missing'] = (data['annual_income'].isnull().astype(int))

# -- Random Sample Imputation
missing = data['annual_income'].isnull()

data.loc[missing, 'annual_income'] = np.random.choice(
    data.loc[~missing, 'annual_income'],
    size=missing.sum()
)

# AFTER
print("\nMissing values AFTER:")
print(data['annual_income'].isnull().sum())
```

Q 7 : Outlier Handling

```python
data.dtypes
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

columns = ['age','annual_income','loan_amount','credit_score','repayment_history','transaction_count','spending_ratio']

fig,axes = plt.subplots(4,2,figsize=(14,17))

for col, ax in zip(columns, axes.flatten()):
    sns.boxplot(x=data[col], ax=ax)
    ax.set_title(f'{col} - Boxplot')

plt.tight_layout()
plt.show()
```

```python
# Z-score [ annul_income ]
from scipy.stats import zscore
# Before Outliers
sns.boxplot(x=data['annual_income'])
plt.title('Befor : annual_income')
plt.show()

# Calculate OUTLIER
data['annual_income_z'] = zscore(data['annual_income'])

# identify outliers
outliers = data[(data['annual_income_z'].abs() > 3)]
print("Total Outliers",len(outliers))

# delete outliers
del_outliers = data[(data['annual_income_z'].abs()) <= 3].copy()

# after remove outliers 
sns.boxplot(x=del_outliers['annual_income_z'])
plt.title("After Z-score")
plt.show()

# BEFORE vs AFTER DIFFERENCE
# ============================================

print("\n----- COMPARISON -----")

print("----- BEFORE Z-SCORE -----")
print(data['annual_income'].describe())

print("\n----- AFTER Z-SCORE -----")
print(del_outliers['annual_income_z'].describe())
```

```python
# IQR Method [ loan_amount ]

# befor graph
sns.boxplot(x=data['loan_amount'])
plt.title('Before_loan_amount')
plt.show()

# IQR method
Q1 = data['loan_amount'].quantile(0.25)
Q3 = data['loan_amount'].quantile(0.75)
IQR = Q3 - Q1
print(f"IQR is : {IQR}")

# Fences
lower_fences = Q1 - 1.5 * IQR
upper_fences = Q3 + 1.5 * IQR
print(f"lower_fences is : {lower_fences} ")
print(f"upper_fences is : {upper_fences} ")

# outliers
outliers = data[(data['loan_amount'] > upper_fences) | (data['loan_amount'] < lower_fences) ]
print("total outliers : ",len(outliers))

# delete outliers
del_outliers = data[(data['loan_amount'] >= lower_fences) & (data['loan_amount'] <= upper_fences)].copy()
print("after remove outliers : ",len(del_outliers))

# after IQR METHOD
sns.boxplot(x = del_outliers['loan_amount'])
plt.title("After IQR loan_amount ")
plt.show()

print("\n----- COMPARISON -----")

print("----- BEFORE IQR -----")
print(data['loan_amount'].describe())

print("\n----- AFTER IQR -----")
print(del_outliers['loan_amount'].describe())
```

```python
# Percentile method [ credit_score ]
import seaborn as sns 
import matplotlib.pyplot as plt
# befor graph
sns.boxplot(x=data['credit_score'])
plt.title('Before Percentile method')
plt.show()

lower = data['credit_score'].quantile(0.01)
upper = data['credit_score'].quantile(0.99)
print('lower outliers : ',lower)
print('upper outliers :',upper)

# Identify outliers
outliers = data[(data['credit_score'] < lower) | (data['credit_score'] > upper)]
print("total outliers",len(outliers))

# removie outliers
removie_outliers = data[(data['credit_score'] >= lower) & (data['credit_score'] <= upper)].copy()

# After remove outliers
sns.boxplot(x=removie_outliers['credit_score'])
plt.title('After Percentile method')
plt.show()
```

```python
# winsorization method [ spending_ratio ]
from scipy.stats.mstats import winsorize

# winsorization use
data['spending_ratio_wins'] = winsorize(data['spending_ratio'],limits=[0 , 0.05])

# Before VS After
print("----- BEFORE WINSORIZATION -----")
print(data['spending_ratio'].describe())

print("\n----- AFTER WINSORIZATION -----")
print(data['spending_ratio_wins'].describe())


# Before graph
plt.figure(figsize=(8,4))
sns.boxplot(x=data['spending_ratio'])
plt.title("Before Winsorization - spending_ratio")
plt.show()

# After graph
plt.figure(figsize=(8,4))
sns.boxplot(x=data['spending_ratio_wins'])
plt.title("After Winsorization - spending_ratio")
plt.show()
```

```python
# Log Transform
from sklearn.preprocessing import FunctionTransformer

log_transformer = FunctionTransformer(
    np.log1p
)

df["spending_ratio_log"] = log_transformer.transform(
    df[["spending_ratio"]]
)

display(
    df[
        ["spending_ratio", "spending_ratio_log"]
    ].head(10)
)
```

```python
# Reciprocal Transform 
reciprocal_transformer = FunctionTransformer(
    lambda x: 1 / (x + 1)
)

df["spending_ratio_reciprocal"] = reciprocal_transformer.transform(
    df[["spending_ratio"]]
)

display(
    df[
        ["spending_ratio", "spending_ratio_reciprocal"]
    ].head(10)
)
```

```python
# Squre root
sqrt_transformer = FunctionTransformer(
    np.sqrt
)

df["spending_ratio_sqrt"] = sqrt_transformer.transform(
    df[["spending_ratio"]]
)

display(
    df[
        ["spending_ratio", "spending_ratio_sqrt"]
    ].head(10)
)
```

```python
# ColumnTransformer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

numeric_columns = [
    "age",
    "annual_income",
    "loan_amount",
    "credit_score",
    "repayment_history",
    "transaction_count",
    "spending_ratio"
]

categorical_columns = [
    "gender",
    "region",
    "education_level",
    "employment_type",
    "loan_purpose"
]

df_ct = df.copy()

# Numerical missing values
df_ct[numeric_columns] = df_ct[numeric_columns].fillna(
    df_ct[numeric_columns].median()
)

# Categorical missing values
df_ct[categorical_columns] = df_ct[categorical_columns].fillna(
    df_ct[categorical_columns].mode().iloc[0]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", StandardScaler(), numeric_columns),
        ("categorical", OneHotEncoder(handle_unknown="ignore"), categorical_columns)
    ]
)

# transformation apply
transformed_data = preprocessor.fit_transform(df_ct)

print("Original shape:", df_ct.shape)
print("Transformed shape:", transformed_data.shape)
```

Numerical, categorical, binary, ordinal aur ID columns identify kiye.

Missing values ko check kiya aur numerical data ke liye Median Imputation aur categorical data ke liye Most Frequent Imputation use kiya.

### 4. Missing Indicator

Missing values ki original information preserve karne ke liye Missing Indicator ka concept use kiya.

### 5. MICE Imputation

Related numerical features ke missing values ko other features ki help se estimate karne ke liye MICE/Iterative Imputation use kiya.

Numerical features mein unusual values identify karne ke liye Z-Score aur IQR methods use kiye.

Identified outliers ko dataset ki requirement ke according remove ya Winsorization ke through treat kiya.

`education_level` jaise ordered categorical feature ko meaningful numerical order mein convert karne ke liye Ordinal Encoding use kiya.

`region` aur `loan_purpose` jaise nominal categorical features ko separate 0/1 columns mein convert karne ke liye One-Hot Encoding use kiya.

`annual_income` ko approximately equal-sized groups mein divide karke `Low`, `Medium`, `High`, aur `Very High` income groups create kiye.

### 12. Binarization

Threshold-based binary features create kiye, jisme condition true hone par `1` aur false hone par `0` assign kiya.
